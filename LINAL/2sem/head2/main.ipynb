```python
from sympy import *
from pprint import pprint
```

```python
C, A = Matrix([[3, 1, 0], [1, 3, 0], [2, 1, 5]]).jordan_form()

for i in range(3):
    A[i, i] = E**A[i, i]

eA = C @ A @ C.inv()
print("diag e^A =", A)
print("C", C)
print("C^(-1)", C.inv())
print("e^A", eA)
```

```python
C, A_J = Matrix([[2, 0, -1], [1, 2, 2], [0, 0, 3]]).jordan_form()
print("A_J =   ", A_J)
print("C =     ", C)
print("C^(-1) =", C.inv())
# print("e^A", eA)
```

```python
lam = Symbol('L')
A = Matrix([[lam, 1, 0], [0, lam, 1], [0, 0, lam]])
print(A**2)
print(A**3)
print(A**4)
```
