```python
fmt = lambda x: f"{int(x // 3600)}°{int(x % 3600 // 60)}'{x % 60}\" ({int(x // 3600)}^\\circ{int(x % 3600 // 60)}'{x % 60}'')"
sec = lambda g, m, s: 3600 * g + 60 * m + s
```

```python
delta_phi_inst = 2 / 3 * 30
```

прямые многократные измерения

```python
g1, m1, s1 = 8, 22, 30
g2, m2, s2 = 8, 21, 0
g3, m3, s3 = 8, 21, 0

phi1 = sec(g1, m1, s1)
phi2 = sec(g2, m2, s2)
phi3 = sec(g3, m3, s3)

phi = (phi1 + phi2 + phi3) / 3
print(fmt(phi))

sigma = (((phi1-phi)**2 + (phi2-phi)**2 + (phi3-phi)**2) / 3)**0.5
print(sigma)
delta_phi_random = 4.3 * sigma
print(delta_phi_random)
delta_phi = (delta_phi_random**2 + delta_phi_inst**2)**0.5

print(delta_phi, fmt(delta_phi))

print(fmt(180))
```

Рассчитать угол дифракции для каждой линии
