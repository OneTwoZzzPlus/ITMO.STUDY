```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('hypothesis_2.csv', header=None)
sample = df.iloc[:, 0].to_numpy()
```

```python
Fs = np.cumsum(np.bincount(sample)) / len(sample)
print(Fs)
```

```python
F1 = lambda t: (0 if t < 0 else t/10) if t <= 10 else 1
```

```python
sup = float('-inf')
for i in range(len(Fs) - 1):
  s = abs(max(F1(i) - Fs[i], F1(i) - Fs[i + 1]))
  sup = max(sup, s)
print(sup)
```

```python
rho = np.sqrt(len(sample)) * sup
print(rho)
```
