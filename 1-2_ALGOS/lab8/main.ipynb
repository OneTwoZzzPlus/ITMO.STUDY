1. Алгоритмы сортировки 
- Tim Sort - Массивы, которые ранее уже сортировались, но были дополнены новыми элементами.
- Counting Sort - Подсчитывает количество элементов с определёнными ключами.
- Heap Sort - Быстро получить наибольшие значения за O(log n)

```python
# Задача 1
import sys

# Инициализация основной коллекции
main_blocks = []
initial = list(map(int, sys.stdin.readline().split()))
if initial:
    main_blocks.append(initial)

n = int(sys.stdin.readline())
minrun = 2  # Фиксированный размер блока для простоты

for _ in range(n):
    # Чтение пакета и разбиение на блоки
    packet = list(map(int, sys.stdin.readline().split()))
    i = 0
    while i < len(packet):
        block = packet[i:i + minrun]
        main_blocks.append(block)
        i += minrun

    # Проверка и слияние блоков
    changed = True
    while changed:
        changed = False
        i = 0
        while i < len(main_blocks) - 1:
            curr_size = len(main_blocks[i])
            next_size = len(main_blocks[i + 1])
            if curr_size > next_size:
                # Слияние блоков и сортировка
                merged = sorted(main_blocks[i] + main_blocks[i + 1])
                main_blocks = main_blocks[:i] + [merged] + main_blocks[i + 2:]
                changed = True
                break
            i += 1

    # Вывод статистики
    max_size = max(len(b) for b in main_blocks) if main_blocks else 0
    print(f"{len(main_blocks)} {max_size}")
```

```python
# Задача 2
def sort_flowers(flowers):
    # Подсчёт частот
    frequency = {}
    for num in flowers:
        if num in frequency:
            frequency[num] += 1
        else:
            frequency[num] = 1
    
    # Сортируем ключи: сначала по убыванию частоты, затем по убыванию значения
    sorted_keys = sorted(frequency.keys(), key=lambda x: (-frequency[x], -x))
    
    # Собираем результат
    result = []
    for num in sorted_keys:
        result.extend([num] * frequency[num])
    
    return result

# Чтение ввода и вывод
flowers = list(map(int, input().split()))
sorted_flowers = sort_flowers(flowers)
print(' '.join(map(str, sorted_flowers)))
```

```python
import heapq

def safe_candy_order(arr, n, m):
    safe = [x for x in arr if x % n != 0]
    if not safe:
        return []
    max_heap = [-x for x in safe]
    heapq.heapify(max_heap)
    result = []
    i = 0
    while max_heap and i <= m:
        result.append(-heapq.heappop(max_heap))
        i += 1
    return result

# Чтение ввода
arr = list(map(int, input().split()))
n, m = map(int, input().split)

# Получение результата и вывод
output = safe_candy_order(arr, n, m)
print(' '.join(map(str, output)) if output else '')
```

```python
def find_max_consecutive_row(matrix):
    max_length = -1
    result_row = 0

    for i, row in enumerate(matrix):
        current_len = 1
        row_max = 1
        for j in range(1, len(row)):
            if row[j] == row[j-1] + 1:
                current_len += 1
                row_max = max(row_max, current_len)
            else:
                current_len = 1
        if row_max > max_length:
            max_length = row_max
            result_row = i
    return result_row
```

```python
available = input().strip()
target = input().strip()

target_counts = {}
for c in target:
    target_counts.setdefault(c, []) += 1

available_counts = {}
for c in available:
    available_counts.setdefault(c, []) += 1

missing = []
for char in sorted(target_counts):
    needed = target_counts[char]
    have = available_counts.get(char, 0)
    if have < needed:
        missing.append(f"{char} {needed - have}")

if not missing:
    print("Все буквы присутствуют")
else:
    print("\n".join(missing))
```

```python
def countDistinct(nums, k, p):
    n = len(nums)
    div = [1 if num % p == 0 else 0 for num in nums]
    prefix = [0] * (n + 1)
    for i in range(n):
        prefix[i+1] = prefix[i] + div[i]
    
    unique = set()
    
    for l in range(n):
        for r in range(l, n):
            total = prefix[r+1] - prefix[l]
            if total > k:
                break
            subarray = tuple(nums[l : r+1])
            unique.add(subarray)
    
    return len(unique)

```

```python
def minimal_encoded_length(S, D):
    len_to_subs = {}
    for sub in D:
        len_to_subs.setdefault(len(sub), set()).add(sub)
    lengths = list(len_to_subs.keys())
    n = len(S)
    dp = [float('inf')] * (n + 1)
    dp[0] = 0
    
    for i in range(1, n + 1):
        dp[i] = dp[i-1] + 1
        for l in lengths:
            if l > i:
                continue
            if S[i-l:i] in len_to_subs[l]:
                dp[i] = min(dp[i], dp[i-l] + 1)
    return dp[n]

```

```python
def find_council_members(arr):
    count1, count2, count3 = 0, 0, 0
    cand1 = cand2 = cand3 = None
    
    for num in arr:
        if num == cand1:
            count1 += 1
        elif num == cand2:
            count2 += 1
        elif num == cand3:
            count3 += 1
        elif count1 == 0:
            cand1, count1 = num, 1
        elif count2 == 0:
            cand2, count2 = num, 1
        elif count3 == 0:
            cand3, count3 = num, 1
        else:
            count1 -= 1
            count2 -= 1
            count3 -= 1
    
    # Проверяем частоты конфет
    threshold = len(arr) // 4
    result = []
    candidates = {cand1, cand2, cand3}
    for cand in candidates:
        if cand is not None and arr.count(cand) > threshold:
            result.append(cand)
    
    result = sorted(result) if result else [-1]
    return result if result[0] != -1 else [-1]

# Ввод и вывод
arr = list(map(int, input().split()))
output = find_council_members(arr)
print(' '.join(map(str, output)) if output != [-1] else -1)
```
