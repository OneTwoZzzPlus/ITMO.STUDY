```python
def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2
        left_half = arr[:mid]
        right_half = arr[mid:]
        
        merge_sort(left_half)
        merge_sort(right_half)
        
        i = j = k = 0
        
        # Слияние отсортированных половин
        while i < len(left_half) and j < len(right_half):
            if left_half[i] < right_half[j]:
                arr[k] = left_half[i]
                i += 1
            else:
                arr[k] = right_half[j]
                j += 1
            k += 1
        
        # Добавление оставшихся элементов из левой половины
        while i < len(left_half):
            arr[k] = left_half[i]
            i += 1
            k += 1
        
        # Добавление оставшихся элементов из правой половины
        while j < len(right_half):
            arr[k] = right_half[j]
            j += 1
            k += 1

# Пример использования
if __name__ == "__main__":
    example_arr = [12, 11, 13, 5, 6, 7]
    print("Исходный массив:", example_arr)
    
    merge_sort(example_arr)
    print("Отсортированный массив:", example_arr)
```

```python
def findLIS(a: list[int]) -> list[int]:
   n = len(a)
   prev = [0] * n
   d = [0] * n
 
   for i in range(n):
       d[i] = 1
       prev[i] = -1
       for j in range(i):
           if (a[j] < a[i] and d[j] + 1 > d[i]):
               d[i] = d[j] + 1
               prev[i] = j
 
   pos = 0
   length = d[0]
   for i in range(n):
       if d[i] > length:
           pos = i
           length = d[i]
   
   answer = []
   while pos != -1:
       answer.append(a[pos])
       pos = prev[pos]
   answer.reverse()
 
   return answer  
```

```python
def prefix_func(s):
    pi = [0] * len(s)
    k = 0
    for i in range(1, len(s)):
        while k > 0 and s[k] != s[i]:
            k = pi[k-1]
        if s[k] == s[i]:
            k += 1
        pi[i] = k
    return pi

print(prefix_func("ababa"))
```

```python
from collections import deque

def bfs(graph, start):
    visited = set()      # Множество посещенных вершин
    queue = deque()      # Очередь для обработки вершин (FIFO)
    result = []          # Результат обхода
    
    queue.append(start)  # Добавляем стартовую вершину в очередь
    visited.add(start)   # Помечаем как посещенную
    
    while queue:
        vertex = queue.popleft()  # Извлекаем вершину из начала очереди
        result.append(vertex)     # Добавляем в результат
        
        # Перебираем всех соседей текущей вершины
        for neighbor in graph[vertex]:
            if neighbor not in visited:
                visited.add(neighbor)      # Помечаем соседа как посещенного
                queue.append(neighbor)     # Добавляем в конец очереди
    
    return result
```
