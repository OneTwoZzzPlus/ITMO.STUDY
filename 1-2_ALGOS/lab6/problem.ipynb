1. Дана матрица размерности X на Y. Дано некоторое целое число k. Вы начинаете в верхней-левой, хотим в нижнюю-правую. В любой момент вы можете спуститься вниз. На нечётных строках вы можете смещаться вправо максимум на k позиций. На чётных строках - влево максимум на k позиций. На последней строке можно смещаться и вправо и влево на любое расстояние. Нужно составить такой алгоритм, который найдёт такую траекторию, для которой сумма чисел, по которым пройдём максимальны.
Собираются все промежуточные клетки, клетка собирается 1 раз.

```python
x, y, k = 3, 3, 3
# mat = [[randint(-100, 100) for _ in range(x)] for _ in range(y)]
# mat = [[-1, -1, -1], [-1, -1, -1], [-1, -1, -1]]
mat = [[1, 1, 1], [1, 1, 1], [1, 1, 1]]
dp = [[float('-inf') for _ in range(x)] for _ in range(y)]

# Последняя строка
dp[-1][-1] = mat[-1][-1]
for i in range(x - 2, -1, -1):
    # Все возможные пути влево и обратно
    left = max(sum(mat[-1][k:i]) for k in range(0, i)) if i != 0 else 0
    # + путь до конечной
    dp[-1][i] = sum(mat[-1][i: x]) + (left if left > 0 else 0)
    
# Остальные
for raw in range(y - 2, -1, -1):
    if raw % 2: # чёт
        for i in range(x):
            # Выбираем, до какой ячейки из k ячеек слева лучше идти 
            left = max(sum(mat[raw][p + 1:i]) + dp[raw][p] for p in range(max(0, i-k), i)) if i != 0 else float('-inf')
            dp[raw][i] = mat[raw][i] + max(dp[raw + 1][i], left)
    else: # нечет
        for i in range(x - 1, -1, -1):
            right = max(sum(mat[raw][i + 1:p]) + dp[raw][p] for p in range(min(x - 1, i+k), i, -1)) if i != x-1 else float('-inf')
            dp[raw][i] = mat[raw][i] + max(dp[raw + 1][i], right)


pprint(mat)
pprint(dp)
print(dp[0][0])
```
