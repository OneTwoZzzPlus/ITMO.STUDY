1)	Пользователю необходимо дать сдачу N рублей. У него имеется M1 монет номиналом S1, M2 монет номиналом S2, M3 монет номиналом S3 и M4 монет номиналом S4. Необходимо найти наименьшую комбинацию из заданных монет, которые позволят получить в сумме N.

```python
# [(Mi, Si), ...], N
MS = [(4, 1), (1, 5), (4, 10), (1, 50)]
N = 37

# Сортируем по убыванию номиналов: O(n*logn) (if n=4 => O(1))
MS.sort(key=lambda x: x[1], reverse=True)

# Перебираем номиналы: O(n)
coins = []
for (M, S) in MS:
    # Берём монет номинала столько, сколько можно вычесть из общей суммы
    # Если монет не хватает, берём все доступные
    mon = min(N // S, M)
    N -= mon*S
    # Добавляем выбранные монеты в ответ
    if mon != 0:
        coins.append((mon, S))
    # Собрали нужную сумму
    if N == 0:
        break

print(f'Потребовалось монет: {len(coins)}')
print(', '.join([f'{m} монет номиналом {s}' for m, s in coins]))
if N != 0:
    print(f'Невозможно подобрать ещё {N} руб.')
```

Вам дан массив ЦП tasks, каждый из которых обозначен буквой от A до Z и числом n. Каждый интервал ЦП может простаивать или позволять завершить одну задачу. Задачи можно выполнять в любом порядке, но есть ограничение: между двумя задачами с одинаковыми метками должен быть промежуток не менее n интервалов.

Возвращает минимальное количество интервалов работы процессора, необходимых для выполнения всех задач.

```python
from typing import List


class Solution:
    def maxProfit(self, prices: List[int], fee: int) -> int:
        no = 0
        yes = float('-inf')
        
        for p in prices:
            # Если к концу дня будут акции: мы купили их ранее ИЛИ сегодня
            yes = max(yes, no - p)
            # Если к концу дня не будет акций: мы продали их ранее ИЛИ сегодня
            no = max(no, yes + p - fee)   
            
        return max(yes, no)
        
print(Solution().maxProfit([1,3,2,8,4,9], 2))
```

4)	Реализовать алгоритм Дейкстры на основе реальных данных по одной из тем. Улицы Санкт-Петербурга.

```python
name_stations = {'Площадь Александра Невского': 0,
                 'Лиговский проспект': 1,
                 'Владимирская': 2, 'Достоевская': 2,
                 'Невский проспект': 3, 'Гостинный двор': 3,
                 'Садовая': 4, 'Сенная площадь': 4, 'Спасская': 4,
                 'Технологический институт': 5,
                 'Звенигородская': 6, 'Пушкинская': 6,
                 'Маяковская': 7, 'Площадь Восстания': 7}
graph = [[0, 32, 0, 0, 0, 0, 0, 20],
         [32, 0, 11, 0, 0, 0, 0, 0],
         [0, 11, 0, 0, 24, 0, 15, 12],
         [0, 0, 0, 0, 15, 0, 0, 16],
         [0, 0, 24, 15, 0, 15, 20, 0],
         [0, 0, 0, 0, 15, 0, 15, 0],
         [0, 0, 15, 0, 20, 15, 0, 0],
         [20, 0, 12, 16, 0, 0, 0, 0]]


def alg_d(st, end):
    global graph
    N = len(graph)
    used = [0, 0, 0, 0, 0, 0, 0, 0]
    d = [float('inf') for i in range(N)]
    d[st] = 0
    for i in range(N):
        v = 0
        for j in range(N):
            if not used[j] and (v == 0 or d[j] < d[v]):
                v = j
        if d[v] is float('inf'):
            break
        used[v] = 1
        for n in range(N):
            if graph[v][n]:
                if d[v] + graph[v][n] < d[n]:
                    d[n] = d[v] + graph[v][n]
    return d[end]

print(alg_d(name_stations['Маяковская'], name_stations['Сенная площадь']))
```

```python
s = input('Откуда?')
e = input('Куда?')
print(alg_d(name_stations[s], name_stations[e]))
```
