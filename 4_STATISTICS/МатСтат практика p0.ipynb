```python
import numpy as np
import matplotlib.pyplot as plt
```

Выборочная медиана называется сатистика

$$med_n = \left\{
\begin{array}{ll}
X_{\left(\frac{n+1}2\right)}, \text{если } n\not\vdots2\\
\frac{X_{\left(\frac{n}2\right)} + X_{\left(\frac{n}2 + 1\right)}}2
\end{array}
\right.$$

Выборочная квантиль

$$Q_n^\alpha = \left\{
\begin{array}{ll}
X_{\left([n\alpha] + 1\right)}, \text{если } na\not\in\mathbb{Z}\\
\frac{X_{\left(\frac{n}2\right)} + X_{\left(\frac{n}2 + 1\right)}}2, \text{если } na\in\mathbb{Z}
\end{array}
\right.$$

```python
sample = np.array([-1, 0, 2, -1, 3, 0, 1, 4, 1, 2, 5])
```

```python
x_sorted = np.sort(sample)
x_unique, counts = np.unique(x_sorted, return_counts=True)
y_unique = np.cumsum(counts) / len(sample)

x_min = x_sorted[0] - 1
x_max = x_sorted[-1] + 1
y_min = -0.05
y_max = 1.05

plt.figure(figsize=(8, 5))

# Горизонтальная линия слева от первого значения (F = 0)
plt.hlines(y=0, xmin=x_min, xmax=x_unique[0], colors='blue', linewidth=2)

# Горизонтальные линии между уникальными значениями
for i in range(len(x_unique) - 1):
    plt.hlines(y=y_unique[i], xmin=x_unique[i], xmax=x_unique[i+1], colors='blue', linewidth=2)

# Горизонтальная линия после последнего значения (F = 1)
plt.hlines(y=1, xmin=x_unique[-1], xmax=x_max, colors='blue', linewidth=2, label='Эмпирическая функция распределения')

# Точки в уникальных значениях (значение функции справа)
plt.scatter(x_unique, y_unique, color='red', zorder=5, label='Значения в узлах')

plt.xlabel('Значение')
plt.ylabel('F(x)')
plt.title('Эмпирическая функция распределения (без вертикальных линий)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)

plt.show()
```
