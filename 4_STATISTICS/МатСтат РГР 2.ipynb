```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
df = pd.read_csv("https://drive.usercontent.google.com/u/0/uc?id=1SoZFjpYau-MDheMYqKidu4d2caMw0IpG&export=download", sep=';')
```

```python
X1 = df["X1"].dropna().to_numpy()
X2 = df["X2"].dropna().to_numpy()
X3 = df["X3"].dropna().to_numpy()
X4 = df["X4"].dropna().to_numpy()

alpha = 0.05
```

# 4.2. Проверка гипотезы о равенстве математических ожиданий для двух независимых выборок

```python
t_stat, p_value = stats.ttest_ind(X1, X2)
print(f'Наблюдаемое значение статистики критерия: {t_stat}')
```

```python
m = len(X1)
n = len(X2)

mean1 = np.mean(X1)
mean2 = np.mean(X2)

var1 = np.var(X1, ddof=1)
var2 = np.var(X2, ddof=1)

# объединённая дисперсия
sp2 = ((m - 1)*var1 + (n - 1)*var2) / (m + n - 2)

t_stat = (mean1 - mean2) / np.sqrt(sp2 * (1/m + 1/n))

print("t-статистика:", t_stat)
```

**1. Гипотезы:**

$$H_0: E(X1) = E(X2)$$
$$H_1: E(X1) \ne E(X2)$$

```python
print(f"2. Уровень значимости: {alpha}")
```

**3. Критерий:**
параметрический t-критерий Стьюдента для двух независимых выборок.
Обоснование:
1) выборки независисы;
2) допущение: выборки приблизительно нормальны

**4. Ошибка первого рода:**
Отвергнуть гипотезу ($H_0$), хотя на самом деле математические ожидания равны.


```python
print(p_value)
print(f"5. Вывод: {"Отвергаем H0" if p_value < alpha else "Нет оснований отвергнуть H0"}")
```

# 4.3. Проверка гипотезы о параметре нормального распределения

```python
sigma2_0 = 14.53
n = 110
```

```python
s2 = np.var(X3, ddof=1) # не смещённая дисперсия

chi2_stat = (n - 1) * s2 / sigma2_0 # критерий дан по заданию; применим потому что столбец подчиняется нормальному распределению

p_left = stats.chi2.cdf(chi2_stat, df=n - 1)
p_right = 1 - stats.chi2.cdf(chi2_stat, df=n - 1)
p_value = 2 * min(p_left, p_right)
p_value = min(p_value, 1.0)

print(f"h = {chi2_stat}")
```

**1. Гипотезы:**

$$H_0: \sigma^2 = 14.53$$
$$H_1: \sigma^2 \ne 14.53$$

```python
print(f"2. Уровень значимости: {alpha}")
```

**3. Критерий:**
Критерий хи-квадрат ($χ²$) для проверки гипотезы о дисперсии нормального распределения.

**4. Ошибка первого рода:**
Отвергнуть гипотезу ($H_0$), хотя на самом деле дисперсия равна 14.53.

```python
print(p_value)
print(f"5. Вывод: {"Отвергаем H0" if p_value < alpha else "Нет оснований отвергнуть H0"}")
```

# 4.4. Непараметрический критерий для двух выборок

```python
_, p_value = stats.mannwhitneyu(X1, X2, alternative="two-sided")
```

**1. Гипотезы:**

$$H_0: \text{распределения X1 и X2 совпадают}$$
$$H_1: \text{распределения различаются}$$


```python
print(f"2. Уровень значимости: {alpha}")
```

**3. Критерий:**
Критерий Манна–Уитни.

**4. Ошибка первого рода:**
Отвергнуть ($H_0$), хотя распределения выборок на самом деле одинаковы.


```python
print(p_value)
print(f"5. Вывод: {"Отвергаем H0" if p_value < alpha else "Нет оснований отвергнуть H0"}")
```

# 4.5. Критерий согласия

```python
lam = 0.217
```

```python
bins = np.linspace(X4.min(), X4.max(), 6)
observed, edges = np.histogram(X4, bins=bins)
probs = []
for i in range(len(edges) - 1):
    a, b = edges[i], edges[i + 1]
    p = stats.expon.cdf(b, scale=1/lam) - stats.expon.cdf(a, scale=1/lam)
    probs.append(p)
probs = np.array(probs)
expected = len(X4) * probs
chi2_stat = np.sum((observed - expected)**2 / expected)
dfree = len(observed) - 1
p_value = 1 - stats.chi2.cdf(chi2_stat, df=dfree)
```

```python
dfree = len(observed) - 1
print("χ² статистика:", chi2_stat)
print("Степени свободы:", dfree)
```

**1. Гипотезы:**

$$H_0: X4 \sim \text{Exp}(\lambda)$$
$$H_1: X4 \text{ не имеет данного распределения}$$

```python
print(f"2. Уровень значимости: {alpha}")
```

**3. Критерий:**
Критерий согласия Пирсона $χ²$.

**4. Ошибка первого рода:**
Отвергнуть гипотезу ($H_0$), хотя выборка действительно имеет показательное распределение с параметром $\lambda$.


```python
print(p_value)
print(f"5. Вывод: {"Отвергаем H0" if p_value < alpha else "Нет оснований отвергнуть H0"}")
```
