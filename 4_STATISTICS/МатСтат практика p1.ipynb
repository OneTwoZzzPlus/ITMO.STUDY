```python
x, y = [0, 42], [0, 68]

i = 1
tol = 1e-12
while abs(x[i-1] - x[i]) >= tol or abs(y[i-1] - y[i]) >= tol:
  X = [z for z in sample if abs(z-x[i]) < abs(z-y[i])]
  Y = [z for z in sample if abs(z-x[i]) > abs(z-y[i])]
  x.append(sum(X)/len(X))
  y.append(sum(Y)/len(Y))
  i += 1

print(x, y, sep='\n')
print(x[-1], y[-1])
x, y = x[-1], y[-1]
```
