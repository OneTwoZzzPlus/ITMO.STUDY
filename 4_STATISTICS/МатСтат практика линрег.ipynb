$ Y  =\theta_0 + \theta_1 X + \varepsilon$

Имеем значения $X = \{X_1, ..., X_2\}$

$ \Delta(\theta_0, \theta_1) = \sum\limits_{i=1}^n \varepsilon_i^2 = \sum\limits_{i=1}^n (Y_i -\theta_0 - \theta_1 X_i)^2$

$\begin{cases}
\frac{\delta Δ}{\delta \theta_0} = -2 \sum\limits_{i=1}^n (Y_i -\theta_0 - \theta_1 X_i) = -2 (\sum Y_i - n \theta_0 - \theta_1 \sum X_i) = 0\\
\frac{\delta Δ}{\delta \theta_1} = -2 \sum\limits_{i=1}^n (Y_i -\theta_0 - \theta_1 X_i) = -2 (\sum X_i Y_i - \theta_0 \sum X_i - \theta_1 \sum X_i^2) = 0
\end{cases}$

$G =\begin{bmatrix}
\frac{\delta^2 Δ}{\delta \theta_0^2} &
\frac{\delta^2 Δ}{\delta \theta_0 \delta \theta_1} \\
\frac{\delta^2 Δ}{\delta \theta_0 \delta \theta_1} &
\frac{\delta^2 Δ}{\delta \theta_1^2}
\end{bmatrix}
=
\begin{bmatrix}
2n &
2\sum X_i \\
2\sum X_i &
2\sum X_i^2
\end{bmatrix}$

$G$ - положительно определённая, значит, это минимум.

$\begin{cases}
\theta_0 = \frac{\sum Y_i - \theta_1 \sum X_i}{n}\\
\theta_1 = \frac{\sum (X_i - \bar{X})(Y_i - \bar{Y})}{\sum (X_i - \bar{X})^2}
\end{cases}$



Условия Гаусса-Маркова

1) $\varepsilon_i$ - независимые одинаково распределённые СВ
2) $E[\varepsilon_i] = 0$
3) $D[\varepsilon_i] = \sigma^2$
4) $\varepsilon_i ≈ N_{0, \sigma^2}$

Отсюда

- $E[Y_i] = \theta_0 + \theta_1 X_i$
- $Y_i ≈ N_{(\theta_0 + \theta_1 X_i), \sigma^2}$

Метод максимального правдоподобия

$f(Y_i) = \frac{1}{\sqrt{2\pi}\sigma} e^{-\frac{(Y_i - \theta_0 - \theta_1 X_i)^2}{2\sigma^2}}$

$f(Y_1 ... Y_n) = \frac{1}{(\sqrt{2\pi}\sigma)^n} e^{-\frac{\sum\limits_{i=1}^n(Y_i - \theta_0 - \theta_1 X_i)^2}{2\sigma^2}}$

Задача на $max f$ по $\theta$ = задаче на $min \Delta$ по $\theta$.

$f(Y_1 ... Y_n) = \frac{1}{(\sqrt{2\pi}\sigma)^n} e^{-\frac{\Delta(\theta_0, \theta_1)}{2\sigma^2}}$

Задача на $max f$ по $\sigma^2$.

$L = \ln{f(Y_1 ... Y_n)} = -n \ln{\sqrt{2\pi}\sigma} -\frac{\Delta}{2\sigma^2} = -\frac{n}{2}(\ln{2\pi} + \ln\sigma^2) -\frac{\Delta}{2\sigma^2}$

$L' = -\frac{n}{2} \cdot \frac{1}{\sigma^2} + \frac{\Delta}{2\sigma^2}$

$L'=0 ⇒ \sigma^2 = \frac{\Delta}{n}$

$L'' = \frac{n}{2} \cdot \frac{1}{\sigma^3} + \frac{\Delta}{\sigma^3} < 0$

$\sigma^2 = \frac{\Delta}{n}$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([9, 4, 17, 2, 14, 10, 13, 1, 11, 3])
Y = np.array([15, 9, 41, 1, 32, 22, 30, 4, 31, 8])
```

```python
plt.scatter(X, Y)
plt.show()
```

```python
theta_1 = (np.sum((X - np.sum(X)) * (Y - np.sum(Y)))) / (np.sum((X - np.sum(X))**2))
theta_0 = (np.sum(Y) - theta_1 * np.sum(X)) / len(X)
print(f"θ₀ = {theta_0}\nθ₁ = {theta_1}")
```

```python
D = np.sum( (Y - theta_0 - theta_1 * X)**2) / len(X)
print(f"D = {D}")
```
